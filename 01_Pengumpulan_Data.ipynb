Notebook ini merepresentasikan blok **Pengumpulan Data -> Primer (KARHUTSET)** pada diagram metodologi penelitian.

### Deskripsi Tahapan:
1. Memuat dataset primer hasil *scraping* media sosial X (Twitter) terkait isu Kebakaran Hutan dan Lahan (`karhutla_dataset.csv`).
2. Melakukan inspeksi struktur data awal (jumlah baris, kolom, tipe data).
3. Mengecek nilai kosong (*missing values*) dan duplikasi data mentah.
4. Menampilkan sampel teks mentah sebelum masuk ke tahap *Data Cleaning*.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur tampilan kolom pandas agar teks terlihat utuh
pd.set_option('display.max_colwidth', 200)
```

## 1.1 Memuat Dataset Primer (`karhutla_dataset.csv`)

```python
dataset_path = 'karhutla_dataset.csv'
df_raw = pd.read_csv(dataset_path)

print(f'Jumlah total baris data : {df_raw.shape[0]}')
print(f'Jumlah total kolom data : {df_raw.shape[1]}')
print(f'Nama kolom              : {list(df_raw.columns)}')
df_raw.head(10)
```

## 1.2 Inspeksi Kualitas Data Mentah (Missing Values & Duplikasi)

```python
print('=== Informasi Tipe Data ===')
print(df_raw.info())

print('\n=== Jumlah Missing Values ===')
print(df_raw.isnull().sum())

print('\n=== Jumlah Duplikasi ===')
print(f"Duplikasi berdasarkan ID   : {df_raw.duplicated(subset=['id']).sum()}")
print(f"Duplikasi berdasarkan Teks : {df_raw.duplicated(subset=['text']).sum()}")
```

## 1.3 Distribusi Panjang Karakter & Kata pada Data Mentah

```python
df_raw['char_length'] = df_raw['text'].astype(str).apply(len)
df_raw['word_count'] = df_raw['text'].astype(str).apply(lambda x: len(x.split()))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df_raw['char_length'], bins=40, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Distribusi Panjang Karakter (Data Mentah)')
axes[0].set_xlabel('Jumlah Karakter')

sns.histplot(df_raw['word_count'], bins=40, kde=True, ax=axes[1], color='darkorange')
axes[1].set_title('Distribusi Jumlah Kata per Dokumen (Data Mentah)')
axes[1].set_xlabel('Jumlah Kata')

plt.tight_layout()
plt.show()

print(df_raw[['char_length', 'word_count']].describe())
```
